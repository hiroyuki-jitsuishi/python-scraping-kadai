```python

# ==================================================
# 侍テラコヤ 自己紹介文の自動更新
# Google Colab + Selenium
# ==================================================

from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# --------------------------------------------------
# 1. 基本設定
# --------------------------------------------------

PROFILE_URL = "https://terakoya.sejuku.net/account/profile"

NEW_INTRODUCTION = (
    "プログラミング学習中です！"
    "今はスクレイピングに挑戦しています！"
)

wait = WebDriverWait(chrome_driver, 20)

# --------------------------------------------------
# 2. プロフィールページへ移動
# --------------------------------------------------

chrome_driver.get(PROFILE_URL)

print("プロフィールページに移動しました")

# --------------------------------------------------
# 3. 編集ボタンをクリック
# --------------------------------------------------

edit_button = wait.until(
    EC.element_to_be_clickable(
        (By.XPATH, "//button[normalize-space()='編集']")
    )
)

edit_button.click()

print("編集ボタンをクリックしました")

# --------------------------------------------------
# 4. 自己紹介欄を取得
# --------------------------------------------------

wait.until(
    lambda driver: len(
        driver.find_elements(By.TAG_NAME, "textarea")
    ) == 2
)

textareas = chrome_driver.find_elements(
    By.TAG_NAME,
    "textarea"
)

# 1番目のtextareaが自己紹介欄
introduction_input = textareas[0]

# 2番目のtextareaは学習目標欄
# 今回は変更しない

# --------------------------------------------------
# 5. 自己紹介欄の項目名を確認
# --------------------------------------------------

label_check = chrome_driver.execute_script(
    """
    let element = arguments[0];

    for (let i = 0; i < 6; i++) {

        if (!element) break;

        const text = element.innerText || "";

        if (text.includes("自己紹介")) {
            return true;
        }

        element = element.parentElement;
    }

    return false;
    """,
    introduction_input
)

print("自己紹介の項目名確認：", label_check)

if not label_check:
    raise RuntimeError(
        "自己紹介欄を確認できませんでした"
    )

# --------------------------------------------------
# 6. 自己紹介文を変更
# --------------------------------------------------

introduction_input.clear()

introduction_input.send_keys(
    NEW_INTRODUCTION
)

input_check = (
    introduction_input.get_attribute("value")
    == NEW_INTRODUCTION
)

print("自己紹介の入力確認：", input_check)

if not input_check:
    raise RuntimeError(
        "自己紹介文の入力に失敗しました"
    )

print("\n===== 更新前の確認 =====")

print(
    "自己紹介：",
    introduction_input.get_attribute("value")
)

# --------------------------------------------------
# 7. 更新ボタンをクリック
# --------------------------------------------------

update_button = wait.until(
    EC.element_to_be_clickable(
        (By.XPATH, "//button[normalize-space()='更新する']")
    )
)

update_button.click()

print("\n更新ボタンをクリックしました")

wait.until(
    EC.element_to_be_clickable(
        (By.XPATH, "//button[normalize-space()='編集']")
    )
)

print("編集画面が終了しました")

# --------------------------------------------------
# 8. プロフィールページを再読み込み
# --------------------------------------------------

chrome_driver.get(PROFILE_URL)

print("プロフィールページを再読み込みしました")

# --------------------------------------------------
# 9. 保存確認のため編集画面を開く
# --------------------------------------------------

edit_button = wait.until(
    EC.element_to_be_clickable(
        (By.XPATH, "//button[normalize-space()='編集']")
    )
)

edit_button.click()

print("保存確認用の編集画面を開きました")

# --------------------------------------------------
# 10. 保存内容を取得
# --------------------------------------------------

wait.until(
    lambda driver: len(
        driver.find_elements(By.TAG_NAME, "textarea")
    ) == 2
)

textareas = chrome_driver.find_elements(
    By.TAG_NAME,
    "textarea"
)

saved_introduction = textareas[0].get_attribute("value")

saved_goal = textareas[1].get_attribute("value")

# --------------------------------------------------
# 11. 保存内容を確認
# --------------------------------------------------

saved_check = (
    saved_introduction == NEW_INTRODUCTION
)

print("\n===== 保存内容の確認 =====")

print("自己紹介：", saved_introduction)
print("学習目標：", saved_goal)

print("自己紹介の保存確認：", saved_check)

# --------------------------------------------------
# 12. 最終結果
# --------------------------------------------------

print("\n================================")

if saved_check:

    print("自己紹介文の自動更新成功")
    print("指定した文章が保存されました")

else:

    print("自己紹介文の保存を確認できませんでした")

print("================================")
```
