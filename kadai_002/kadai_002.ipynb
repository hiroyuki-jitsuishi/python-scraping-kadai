```python
!pip install requests -q
import requests
from getpass import getpass

# Google Maps APIキー入力
api_key = getpass('APIキーを入力してください: ')

# 駅名を入力
search_word = input('駅名を入力してください: ')

# APIで駅の位置を検索
geocoding_url = 'https://maps.googleapis.com/maps/api/geocode/json'

geocoding_params = {
    'address': search_word,
    'key': api_key,
    'language': 'ja'
}

response = requests.get(
    geocoding_url,
    params=geocoding_params
)

response.raise_for_status()

geocoding_data = response.json()

# 駅の位置情報を確認
if geocoding_data['status'] != 'OK':

    print('駅の位置情報を取得できませんでした。')
    print('エラー:', geocoding_data['status'])

else:

    # 駅の緯度・経度を取得
    location = geocoding_data['results'][0]['geometry']['location']

    latitude = location['lat']
    longitude = location['lng']

    # レストラン検索API
    url = 'https://maps.googleapis.com/maps/api/place/nearbysearch/json'

    # 検索条件
    params = {
        'location': f'{latitude},{longitude}',
        'radius': 500,
        'type': 'restaurant',
        'language': 'ja',
        'key': api_key
    }

    # APIへアクセス
    response = requests.get(
        url,
        params=params
    )

    response.raise_for_status()

    data = response.json()

    # APIのステータスを確認
    if data['status'] != 'OK':

        print()
        print('レストラン情報を取得できませんでした。')
        print('APIエラー:', data['status'])

    else:

        # レストラン情報を取得
        restaurants = data.get('results', [])
        # レストラン情報を表示
        for i, restaurant in enumerate(restaurants, start=1):

            # レストラン名
            name = restaurant.get(
                'name',
                '情報なし'
            )

            # 評価
            rating = restaurant.get(
                'rating',
                '評価なし'
            )

            # 住所
            address = restaurant.get(
                'vicinity',
                '住所なし'
            )

            print()
            print(f'レストラン名：{name}')
            print(f'評価：{rating}')
            print(f'住所：{address}')
```
